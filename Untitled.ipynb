```python
import numpy as np
import matplotlib.pyplot as plt

T = 50.0
dt = 0.01
t = np.arange(0, T, dt)

def I_ext(t):
    return 10.0 if 5 < t < 30 else 0.0

C_m = 1.0

g_Na = 120.0
g_K = 36.0
g_L = 0.3

E_Na = 50.0
E_K = -77.0
E_L = -54.4

def alpha_n(V):
    return 0.01 * (V + 55) / (1 - np.exp(-(V + 55) / 10))

def beta_n(V):
    return 0.125 * np.exp(-(V + 65) / 80)

def alpha_m(V):
    return 0.1 * (V + 40) / (1 - np.exp(-(V + 40) / 10))

def beta_m(V):
    return 4.0 * np.exp(-(V + 65) / 18)

def alpha_h(V):
    return 0.07 * np.exp(-(V + 65) / 20)

def beta_h(V):
    return 1 / (1 + np.exp(-(V + 35) / 10))

V = np.zeros(len(t))
m = np.zeros(len(t))
h = np.zeros(len(t))
n = np.zeros(len(t))

V[0] = -65

m[0] = 0.05
h[0] = 0.6
n[0] = 0.32

for i in range(len(t) - 1):

    INa = g_Na * (m[i]**3) * h[i] * (V[i] - E_Na)
    IK = g_K * (n[i]**4) * (V[i] - E_K)
    IL = g_L * (V[i] - E_L)

    dV = (I_ext(t[i]) - INa - IK - IL) / C_m
    dm = alpha_m(V[i]) * (1 - m[i]) - beta_m(V[i]) * m[i]
    dh = alpha_h(V[i]) * (1 - h[i]) - beta_h(V[i]) * h[i]
    dn = alpha_n(V[i]) * (1 - n[i]) - beta_n(V[i]) * n[i]

    V[i+1] = V[i] + dt * dV
    m[i+1] = m[i] + dt * dm
    h[i+1] = h[i] + dt * dh
    n[i+1] = n[i] + dt * dn

plt.figure(figsize=(10, 5))
plt.plot(t, V)
plt.title("Hodgkin–Huxley Model (Euler Method)")
plt.xlabel("Time (ms)")
plt.ylabel("Membrane potential V (mV)")
plt.grid()
plt.show()
```
