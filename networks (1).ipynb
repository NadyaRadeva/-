```python
import numpy as np
import matplotlib.pyplot as plt

def I(t):
    if t <= 30 and t >= 0:
        return 30
    else:
        return 0

def normalization(current):
    current = np.array([I(t) for t in range(30)])
    mini = np.argmin(current)
    maxi = np.argmax(current)
    
    normalization = np.zeros(current.size())
    for i in range(current.size()):
        normalization[i] = (current[i] - mini)/(maxi - mini)
    
    return normalization

# ti -the time i when the spike occurs
def input_function(ti):
    A = 55
    values = []
    tau = 0.0001
    time = np.linspace(0, 10, 100)
    for t in time:
        values.append(A*np.exp(-abs(t - ti)/tau))
    plt.plot(time, values)
    
    # print(values)
    return values

# input_function()

def sum_spikes():
    suma = np.zeros([100])
    time = np.linspace(0, 10, 100)
    
    for i in range(10):
        suma += input_function(i)
    # plt.plot(time, input_function(i))
        plt.show()
    # plt.plot(time, input_function(i))
    # plt.show()
    return suma

sum_spikes()

```
