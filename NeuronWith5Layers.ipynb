```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = 0.02
b = 0.2
c = -65
d = 2
```

```python
def I_input(t):
    if 10 < t < 40:
        return 10
    return 0
```

```python
def dv_dt(v, u, I):
    return 0.04*v**2 + 5*v + 140 - u + I

def du_dt(v, u):
    return a*(v*b-u)
```

```python
def izikevich_model(I):
    v = -65
    u = b*v
    step = 0.01
    t_start = 0
    t_end = 50

    time = np.arange(t_start, t_end, step)
    v_vector = []


    for dt in time:
        I = I_input(dt)
        dV = dv_dt(v,u,I)
        dU = du_dt(v,u)

        v = v + step*dV
        u = u + step*dU

        if v >= 30:
            v = c
            u += d

        v_vector.append(v)

    plt.plot(time,v_vector)
    plt.show()
    return v_vector
```
